# AI Multi-Agent Data Analyst System

A multi-agent AI system that analyzes raw sales data, detects data quality issues, cleans the dataset, plans useful analyses, performs verified calculations with Python, generates business insights, and recommends practical actions.

- Final structured analysis report

## How It Works

The system begins by receiving raw sales data in CSV format.

The **Data Inspector Agent** examines the structure of the dataset and identifies potential data quality problems such as missing values, suspicious data types, invalid values, and duplicates.

Python then performs the actual cleaning operations using predefined and deterministic rules.

The **Analyst Agent** studies the dataset structure and selects useful business analyses from a controlled set of available calculations.

The **Python Analysis Engine** performs all numerical calculations to ensure that important business metrics are accurate and reproducible.

The **Insight Agent** interprets the verified results and converts them into understandable business observations.

The **Recommendation Agent** uses those insights to suggest practical business actions.

Finally, the system combines the outputs from the full workflow into one structured data analysis report.

## Agents

### Data Inspector Agent

Analyzes the dataset profile and identifies potential data quality issues.

It examines:

- Columns
- Data types
- Missing values
- Duplicate rows
- Suspicious values
- Possible cleaning requirements

### Analyst Agent

Determines which analyses are most useful for understanding the dataset.

The agent does not perform numerical calculations itself.

Instead, it selects analyses that Python should execute.

### Insight Agent

Reads the verified Python analysis results and converts them into business insights.

It identifies:

- Strong performance
- Weak performance
- Important trends
- Relevant observations

### Recommendation Agent

Converts verified analysis results and insights into practical business recommendations.

Each recommendation is connected directly to the available data.

## Python Tools

### Data Cleaning Tool

Python performs deterministic data cleaning operations.

Examples include:

- Removing duplicate rows
- Converting invalid dates into missing values
- Filling missing categories with `Unknown`
- Calculating missing revenue when both quantity and price are available
- Preserving unresolved missing values instead of inventing data

### Analysis Engine

Python performs all important numerical calculations.

Supported analyses include:


## AI + Python Design Principle

A key architectural principle in this project is separating AI reasoning from deterministic computation.

```text
AI
↓
Understand
Plan
Interpret
Recommend

Python
↓
Inspect
Clean
Calculate
Verify
```

AI is useful for understanding context and interpreting results.

Python is used when accuracy and deterministic calculations are required.

## Data Quality Principle

The system follows an important data analysis rule:

```text
Missing Value != Zero
```

A missing value represents unknown information.

A zero represents a known value of zero.

The system therefore avoids replacing missing business data with zero unless there is evidence that zero is the correct value.

Another principle used throughout the workflow is:

```text
Fix what you can prove.
Flag what you cannot.
```

````python
# =========================================
# CREATE SAMPLE SALES DATA
# =========================================

import pandas as pd

data = {
    "Date": [
        "2026-01-05",
        "2026-01-10",
        "2026-02-03",
        "2026-02-15",
        "2026-03-01",
        "2026-03-12",
        "wrong-date"
    ],

    "Product": [
        "Laptop",
        "Mouse",
        "Laptop",
        "Keyboard",
        "Mouse",
        "Laptop",
        "Keyboard"
    ],

    "Category": [
        "Electronics",
        "Accessories",
        "Electronics",
        "Accessories",
        "Accessories",
        "Electronics",
        None
    ],

    "Quantity": [
        2,
        5,
        1,
        3,
        4,
        None,
        2
    ],

    "Price": [
        1000,
        20,
        1000,
        50,
        20,
        1000,
        50
    ],

    "Revenue": [
        2000,
        100,
        1000,
        150,
        None,
        None,
        100
    ]
}

df = pd.DataFrame(data)

df.to_csv(
    "sample_sales_data.csv",
    index=False
)

print(df)

print("\nCSV created ✅")
````

````python
!pip install -q groq

from groq import Groq
from google.colab import userdata
import json

GROQ_API_KEY = userdata.get("GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

MODEL = "qwen/qwen3.6-27b"

print("Data Analyst AI ready ✅")
````

````python
# =========================================
# DATA INSPECTOR AGENT
# =========================================

def data_inspector_agent(df):

    # Python collects exact facts about the dataset
    data_profile = {
        "rows": len(df),
        "columns": list(df.columns),
        "data_types": df.dtypes.astype(str).to_dict(),
        "missing_values": df.isnull().sum().to_dict(),
        "duplicate_rows": int(df.duplicated().sum()),
        "sample_data": df.head().to_dict(orient="records")
    }

    prompt = f"""
You are a data quality inspection agent.

Analyze this dataset profile:

{data_profile}

Identify:

1. Missing values
2. Suspicious or invalid values
3. Possible data type problems
4. Duplicate rows
5. Columns that may need cleaning

Do not invent problems that are not supported by the data.

Return a short structured report with:

DATASET OVERVIEW
DATA QUALITY ISSUES
RECOMMENDED CLEANING STEPS
"""

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content


inspection_report = data_inspector_agent(df)

print(inspection_report)
````

````python
# =========================================
# DATA CLEANING TOOL
# =========================================

def clean_sales_data(df):

    cleaned_df = df.copy()
    cleaning_log = []

    # 1. Remove duplicate rows
    duplicates = cleaned_df.duplicated().sum()

    if duplicates > 0:
        cleaned_df = cleaned_df.drop_duplicates()
        cleaning_log.append(
            f"Removed {duplicates} duplicate row(s)"
        )

    # 2. Clean Date column
    invalid_dates_before = pd.to_datetime(
        cleaned_df["Date"],
        errors="coerce"
    ).isna().sum()

    cleaned_df["Date"] = pd.to_datetime(
        cleaned_df["Date"],
        errors="coerce"
    )

    if invalid_dates_before > 0:
        cleaning_log.append(
            f"Converted {invalid_dates_before} invalid date(s) to missing values"
        )

    # 3. Fill missing Category
    missing_categories = cleaned_df["Category"].isna().sum()

    cleaned_df["Category"] = cleaned_df["Category"].fillna(
        "Unknown"
    )

    if missing_categories > 0:
        cleaning_log.append(
            f"Filled {missing_categories} missing Category value(s) with 'Unknown'"
        )

    # 4. Calculate missing Revenue when possible
    revenue_missing_before = cleaned_df["Revenue"].isna().sum()

    can_calculate_revenue = (
        cleaned_df["Revenue"].isna()
        & cleaned_df["Quantity"].notna()
        & cleaned_df["Price"].notna()
    )

    cleaned_df.loc[
        can_calculate_revenue,
        "Revenue"
    ] = (
        cleaned_df.loc[can_calculate_revenue, "Quantity"]
        *
        cleaned_df.loc[can_calculate_revenue, "Price"]
    )

    calculated_revenues = can_calculate_revenue.sum()

    if calculated_revenues > 0:
        cleaning_log.append(
            f"Calculated {calculated_revenues} missing Revenue value(s)"
        )

    # 5. Report unresolved missing values
    remaining_missing = cleaned_df.isna().sum()

    unresolved = {
        column: int(count)
        for column, count in remaining_missing.items()
        if count > 0
    }

    if unresolved:
        cleaning_log.append(
            f"Remaining missing values: {unresolved}"
        )

    return cleaned_df, cleaning_log


cleaned_df, cleaning_log = clean_sales_data(df)

print("CLEANED DATA")
print(cleaned_df)

print("\nCLEANING LOG")

for item in cleaning_log:
    print("-", item)
````

````python
# =========================================
# ANALYST AGENT
# =========================================

def analyst_agent(df):

    dataset_info = {
        "rows": len(df),
        "columns": list(df.columns),
        "data_types": df.dtypes.astype(str).to_dict(),
        "missing_values": df.isna().sum().to_dict(),
        "products": df["Product"].dropna().unique().tolist(),
        "categories": df["Category"].dropna().unique().tolist()
    }

    prompt = f"""
You are a business data analyst.

Review this sales dataset:

{dataset_info}

Choose the most useful analyses for understanding business performance.

You may ONLY choose from:

TOTAL_REVENUE
TOTAL_QUANTITY_SOLD
TOP_PRODUCT_BY_REVENUE
TOP_PRODUCT_BY_QUANTITY
REVENUE_BY_PRODUCT
REVENUE_BY_CATEGORY
REVENUE_BY_MONTH
AVERAGE_SALE_VALUE

Return only the analysis names.

One analysis per line.

Do not calculate anything yourself.
Python will perform all calculations.
"""

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content


analysis_plan = analyst_agent(cleaned_df)

print("ANALYSIS PLAN")
print("-------------")
print(analysis_plan)
````

````python
# =========================================
# PYTHON ANALYSIS ENGINE
# =========================================

def run_analysis_engine(df, analysis_plan):

    selected_analyses = [
        line.strip()
        for line in analysis_plan.splitlines()
        if line.strip()
    ]

    results = {}

    # TOTAL REVENUE
    if "TOTAL_REVENUE" in selected_analyses:
        results["TOTAL_REVENUE"] = round(
            df["Revenue"].sum(skipna=True),
            2
        )

    # TOTAL QUANTITY SOLD
    if "TOTAL_QUANTITY_SOLD" in selected_analyses:
        results["TOTAL_QUANTITY_SOLD"] = round(
            df["Quantity"].sum(skipna=True),
            2
        )

    # TOP PRODUCT BY REVENUE
    if "TOP_PRODUCT_BY_REVENUE" in selected_analyses:

        revenue_by_product = (
            df.groupby("Product")["Revenue"]
            .sum()
            .sort_values(ascending=False)
        )

        if not revenue_by_product.empty:
            results["TOP_PRODUCT_BY_REVENUE"] = {
                "product": revenue_by_product.index[0],
                "revenue": round(
                    float(revenue_by_product.iloc[0]),
                    2
                )
            }

    # TOP PRODUCT BY QUANTITY
    if "TOP_PRODUCT_BY_QUANTITY" in selected_analyses:

        quantity_by_product = (
            df.groupby("Product")["Quantity"]
            .sum()
            .sort_values(ascending=False)
        )

        if not quantity_by_product.empty:
            results["TOP_PRODUCT_BY_QUANTITY"] = {
                "product": quantity_by_product.index[0],
                "quantity": round(
                    float(quantity_by_product.iloc[0]),
                    2
                )
            }

    # REVENUE BY PRODUCT
    if "REVENUE_BY_PRODUCT" in selected_analyses:

        results["REVENUE_BY_PRODUCT"] = (
            df.groupby("Product")["Revenue"]
            .sum()
            .round(2)
            .to_dict()
        )

    # REVENUE BY CATEGORY
    if "REVENUE_BY_CATEGORY" in selected_analyses:

        results["REVENUE_BY_CATEGORY"] = (
            df.groupby("Category")["Revenue"]
            .sum()
            .round(2)
            .to_dict()
        )

    # REVENUE BY MONTH
    if "REVENUE_BY_MONTH" in selected_analyses:

        valid_dates_df = df.dropna(
            subset=["Date"]
        ).copy()

        valid_dates_df["Month"] = (
            valid_dates_df["Date"]
            .dt.to_period("M")
            .astype(str)
        )

        results["REVENUE_BY_MONTH"] = (
            valid_dates_df.groupby("Month")["Revenue"]
            .sum()
            .round(2)
            .to_dict()
        )

    # AVERAGE SALE VALUE
    if "AVERAGE_SALE_VALUE" in selected_analyses:

        results["AVERAGE_SALE_VALUE"] = round(
            df["Revenue"].mean(skipna=True),
            2
        )

    return results


analysis_results = run_analysis_engine(
    cleaned_df,
    analysis_plan
)

print("ANALYSIS RESULTS")
print("----------------")

for key, value in analysis_results.items():
    print(f"\n{key}:")
    print(value)
````

````python
# =========================================
# INSIGHT AGENT
# =========================================

def insight_agent(analysis_results):

    prompt = f"""
You are a business intelligence agent.

Here are verified analysis results calculated by Python:

{analysis_results}

Analyze the results and extract useful business insights.

Rules:
- Do not invent numbers.
- Use only the provided results.
- Highlight important trends.
- Mention strong and weak performance.
- Keep the insights practical and concise.

Return the result in this structure:

KEY INSIGHTS

1. ...
2. ...
3. ...

IMPORTANT OBSERVATIONS

- ...
- ...
"""

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content


insights = insight_agent(analysis_results)

print(insights)
````

````python
# =========================================
# RECOMMENDATION AGENT
# =========================================

def recommendation_agent(analysis_results, insights):

    prompt = f"""
You are a business strategy recommendation agent.

Verified analysis results:

{analysis_results}

Business insights:

{insights}

Based only on this information, create practical business recommendations.

Rules:
- Do not invent facts or numbers.
- Recommendations must be directly connected to the analysis.
- Prioritize the most important actions.
- Keep recommendations realistic and practical.
- Mention the reason behind each recommendation.

Return the result in this structure:

BUSINESS RECOMMENDATIONS

1. Recommendation:
   Reason:

2. Recommendation:
   Reason:

3. Recommendation:
   Reason:

PRIORITY ACTION

State the single most important action the business should take first.
"""

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content


recommendations = recommendation_agent(
    analysis_results,
    insights
)

print(recommendations)
````

````python
# =========================================
# FINAL DATA ANALYSIS REPORT
# =========================================

def create_final_report(
    inspection_report,
    cleaning_log,
    analysis_plan,
    analysis_results,
    insights,
    recommendations
):

    report = f"""
=========================================
AI MULTI-AGENT DATA ANALYSIS REPORT
=========================================

1. DATA INSPECTION
------------------
{inspection_report}


2. DATA CLEANING
----------------
"""

    for item in cleaning_log:
        report += f"- {item}\n"

    report += f"""

3. ANALYSIS PLAN
----------------
{analysis_plan}


4. VERIFIED ANALYSIS RESULTS
----------------------------
"""

    for key, value in analysis_results.items():
        report += f"\n{key}:\n{value}\n"

    report += f"""

5. BUSINESS INSIGHTS
--------------------
{insights}


6. BUSINESS RECOMMENDATIONS
---------------------------
{recommendations}


=========================================
END OF REPORT
=========================================
"""

    return report


final_report = create_final_report(
    inspection_report,
    cleaning_log,
    analysis_plan,
    analysis_results,
    insights,
    recommendations
)

print(final_report)
````
